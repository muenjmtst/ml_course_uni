## 1.2 A2. Descriptive Statistics (0.5)

```python
num_values = ['latitude', 'longitude', 'price', 'minimum_nights', 'number_of_reviews','reviews_per_month','calculated_host_listings_count', 'availability_365']
```

I skipped id and host_id because it's just identificator and we don't receive helful info. Also i did not add last_reviews 'cause it's date.

```python
dat[num_values].describe()
```

```python
mode_values = []
var_values = []
iqr_values = [] 
range_values = []
mean_values = []
std_values = []
median_values = []

for items in dat[num_values]:
    var_values.append(dat[items].var())
    range_values.append(dat[items].max() - dat[items].min())
    iqr_values.append(dat[items].quantile(0.75) - dat[items].quantile(0.25))
    mean_values.append(dat[items].mean())
    std_values.append(dat[items].std())
    median_values.append(dat[items].quantile(0.5))
    mode_values.append(dat[items].mode()[0])


result_table = pd.DataFrame({
    'Feature': num_values,
    'Mean': mean_values,
    'Median': median_values,
    'Mode': mode_values,
    'Range': range_values,
    'Variance': var_values,
    'STD': std_values,
    'IQR': iqr_values
})

result_table.round(2)
```

```python
import matplotlib.pyplot as plt

for items in dat[num_values]:
    plt.figure()
    dat[items].plot.box()
    plt.title(f'Boxplot of {items}')
    # plt.xlabel(items)
    plt.ylabel(items)
    plt.show()

```

```python
lower_bound_val = []
upper_bound_val = []
outlier_count = []
first_10_outlier_example = []


for items in dat[num_values]:
    q1 = dat[items].quantile(0.25)
    q3 = dat[items].quantile(0.75)
    iqr = q3 - q1

    lower_bound = q1 - 1.5*iqr
    upper_bound = q3 + 1.5*iqr

    outlier_mask = (dat[items] < lower_bound) | (dat[items] > upper_bound)

    lower_bound_val.append(lower_bound)
    upper_bound_val.append(upper_bound)
    outlier_count.append(outlier_mask.sum())
    first_10_outlier_example.append(dat.loc[outlier_mask,items].head(10).tolist())


outlier_detection_table = pd.DataFrame({
    'Features': num_values,
    'Lower_bound':lower_bound_val,
    'Upper_bound':upper_bound_val,
    'Outlier_count':outlier_count,
    'Outlier_example':first_10_outlier_example
})

outlier_detection_table
```

Only the first 10 outlier values are displayed because some features contain thousands of suspected outliers.

```python
result_table.round(2)
```

## 1.3 A3. Handling Missing Data

```python
reviews_per_month_null = dat['reviews_per_month'].isnull().sum()
rows = dat.shape[0]
```

```python
ratio = (reviews_per_month_null / rows)
print(f'Quantity of null values {ratio * 100:.2f}%')
```

```python
reviews_per_month_missing = (dat['reviews_per_month'].isna())
number_of_reviews_0 = (dat['number_of_reviews'] == 0)

final_cond = reviews_per_month_missing & number_of_reviews_0
final_cond.sum()
```

```python
def display_res(data,name):
    mean = data[name].mean()
    std = data[name].std()
    median = data[name].median()

    print(f'Mean = {mean}')
    print(f'Std = {std} ')
    print(f'Median = {median}')

display_res(dat,'reviews_per_month')
```

### Constant imputation

```python
constant_data = dat.copy()
```

```python
constant_data['reviews_per_month'] = constant_data['reviews_per_month'].fillna(0)
constant_data['reviews_per_month'].isnull().sum()
```

```python
display_res(constant_data, 'reviews_per_month')
```

### Median Imputation

```python
median_data = dat.copy()
```

```python
median = median_data['reviews_per_month'].median()
median_data['reviews_per_month'] = median_data['reviews_per_month'].fillna(median)
median_data['reviews_per_month'].isnull().sum()
```

```python
display_res(median_data, 'reviews_per_month')
```

### Predictive Imputation

```python
predictive_data = dat.copy()
```

```python
predictor_features = ['number_of_reviews', 'price', 'minimum_nights', 'availability_365', 'calculated_host_listings_count']

reviews_per_month_known = (predictive_data['reviews_per_month'].notna())
reviews_per_month_unknown = (predictive_data['reviews_per_month'].isna())

X_train = predictive_data.loc[reviews_per_month_known, predictor_features]
y_train = predictive_data.loc[reviews_per_month_known, 'reviews_per_month']
X_missing = predictive_data.loc[reviews_per_month_unknown, predictor_features]

print(X_train.shape)
print(y_train.shape)
print(X_missing.shape)
```

```python
from sklearn.linear_model import LinearRegression

lin_reg = LinearRegression()

lin_reg.fit(X_train, y_train)
predict = lin_reg.predict(X_missing)

print(predict.shape, predict.max(), predict.min(), predict.mean())
```

```python
(predict < 0).sum()
```

```python
import numpy as np
predict_clipped = np.clip(predict, 0, None)

print(predict_clipped.min())
print((predict_clipped < 0).sum())
print(predict_clipped.mean())
```

```python
predictive_data.loc[reviews_per_month_unknown,'reviews_per_month'] = predict_clipped

print(predictive_data['reviews_per_month'].isna().sum())
display_res(predictive_data, 'reviews_per_month')
```

```python
methods_imp = ['Original', 'Constant(0)', 'Median', 'Predictive']
mean_list = []
median_list = []
std_list = []
missing_values_list = []

imp_list_names = [dat,constant_data, median_data,predictive_data]


def ret_res(data,name):

    mean = data[name].mean()
    std = data[name].std()
    median = data[name].median()
    missing_values = data[name].isnull().sum()

    median_list.append(median)
    std_list.append(std)
    missing_values_list.append(missing_values)
    mean_list.append(mean)



for items in imp_list_names:
    ret_res(items,'reviews_per_month')
     

table_of_comparing = pd.DataFrame({
    'Method':methods_imp,
    'STD':std_list,
    'Median':median_list,
    'Mean':mean_list,
    'Missing_values':missing_values_list
})

table_of_comparing.round(3)
```

### Comparison and Conclusion

- Constant 0: logically the best method for this dataset, since all missing values ​​correspond to listings with no reviews. However, it noticeably lowers the mean and median.
- Median: robust to outliers and preserves the median, but assigns the same value to a large number of rows.
- Predictive: uses information from other features and generates varied values, but is more complex, prone to errors, and initially produced 110 impossible negative values.
- Conclusion: we choose 0 here not because it better preserves statistical metrics, but because it better aligns with the meaning of the data.
