```python
mode_values = []
var_values = []
iqr_values = [] 
range_values = []
mean_values = []
std_values = []
median_values = []

for items in dat[num_values]:
    var_values.append(dat[items].var())
    range_values.append(dat[items].max() - dat[items].min())
    iqr_values.append(dat[items].quantile(0.75) - dat[items].quantile(0.25))
    mean_values.append(dat[items].mean())
    std_values.append(dat[items].std())
    median_values.append(dat[items].quantile(0.5))
    mode_values.append(dat[items].mode()[0])


result_table = pd.DataFrame({
    'Feature': num_values,
    'Mean': mean_values,
    'Median': median_values,
    'Mode': mode_values,
    'Range': range_values,
    'Variance': var_values,
    'STD': std_values,
    'IQR': iqr_values
})

result_table.round(2)
```

```python
import matplotlib.pyplot as plt

for items in dat[num_values]:
    plt.figure()
    dat[items].plot.box()
    plt.title(f'Boxplot of {items}')
    # plt.xlabel(items)
    plt.ylabel(items)
    plt.show()

```

```python
lower_bound_val = []
upper_bound_val = []
outlier_count = []
first_10_outlier_example = []


for items in dat[num_values]:
    q1 = dat[items].quantile(0.25)
    q3 = dat[items].quantile(0.75)
    iqr = q3 - q1

    lower_bound = q1 - 1.5*iqr
    upper_bound = q3 + 1.5*iqr

    outlier_mask = (dat[items] < lower_bound) | (dat[items] > upper_bound)

    lower_bound_val.append(lower_bound)
    upper_bound_val.append(upper_bound)
    outlier_count.append(outlier_mask.sum())
    first_10_outlier_example.append(dat.loc[outlier_mask,items].head(10).tolist())


outlier_detection_table = pd.DataFrame({
    'Features': num_values,
    'Lower_bound':lower_bound_val,
    'Upper_bound':upper_bound_val,
    'Outlier_count':outlier_count,
    'Outlier_example':first_10_outlier_example
})

outlier_detection_table
```

```python
methods_imp = ['Original', 'Constant(0)', 'Median', 'Predictive']
mean_list = []
median_list = []
std_list = []
missing_values_list = []

imp_list_names = [dat,constant_data, median_data,predictive_data]


def ret_res(data,name):

    mean = data[name].mean()
    std = data[name].std()
    median = data[name].median()
    missing_values = data[name].isnull().sum()

    median_list.append(median)
    std_list.append(std)
    missing_values_list.append(missing_values)
    mean_list.append(mean)



for items in imp_list_names:
    ret_res(items,'reviews_per_month')
     

table_of_comparing = pd.DataFrame({
    'Method':methods_imp,
    'STD':std_list,
    'Median':median_list,
    'Mean':mean_list,
    'Missing_values':missing_values_list
})

table_of_comparing.round(3)
```

### Comparison and Conclusion

- Constant 0: logically the best method for this dataset, since all missing values ​​correspond to listings with no reviews. However, it noticeably lowers the mean and median.
- Median: robust to outliers and preserves the median, but assigns the same value to a large number of rows.
- Predictive: uses information from other features and generates varied values, but is more complex, prone to errors, and initially produced 110 impossible negative values.
- Conclusion: we choose 0 here not because it better preserves statistical metrics, but because it better aligns with the meaning of the data.

### 3.1 D1. Measures of Similarity and Dissimilarity (0.5)

```python
import numpy as np

vector_a = np.array([2.3 , 4.5 , 7.8 , 9.1 , 35.6],dtype=float)
vector_b = np.array([3.2 , 5.4 , 8.7 , 1.9 , 5.67],dtype=float)


eu_dist = np.sqrt(np.sum((vector_a - vector_b)**2))

print(eu_dist)
```

```python
set_a = {1, 2, 3, 4, 5}
set_b = {5, 3, 10, 12, 1}

union_set = len(set_a.union(set_b))
intersection_set = len(set_a.intersection(set_b))

jaccard_similarity = intersection_set / union_set
print(jaccard_similarity)
```

```python
cosine_mult = np.dot(vector_a, vector_b)
norm_a = np.linalg.norm(vector_a)
norm_b = np.linalg.norm(vector_b)

cosine_similarity = (cosine_mult) / (norm_a * norm_b)

print(cosine_similarity)
```

```python
mean_a = np.mean(vector_a)
mean_b = np.mean(vector_b)

pearson = 0

first_half =  np.sum((vector_a - mean_a) * (vector_b - mean_b))
second_half = ((np.sum((vector_a- mean_a)**2)) * (np.sum((vector_b - mean_b)**2)))
sqrt_second_half = np.sqrt(second_half)

pearson = first_half / sqrt_second_half


print(pearson)
```

Euclidean distance - just a distance between 2 points, if value of euclidean distance little that means that distance between them low, in another words they looking similar, if 0 then they absolutely similar, high value means big difference. Euclidean distance  can be used with knn , or recommendation systems.

Cosine similarity - compares proportion of values and their direction, 1 - same direction , 0 vectors orthogonal direction - no similarity,  and -1 means opposite direction. 

Pearson correlation - when values of one vector getting higher or bigger, is the same story inside second vector? if pearson = 1 - true, else -1 one getting bigger another is getting smaller, but values can be inside this range -1 to 1, so for example 0.17 means weak positive relation.

Jaccard similarity measures what proportion of all unique elements is shared by both sets.

```python
names = ['Euclidean', 'Cosine', 'Jaccard', 'Pearson']
results = [eu_dist,cosine_similarity, jaccard_similarity, pearson]

res_table = pd.DataFrame({
    'Names': names,
    'Results': results
})

res_table
```

The Euclidean distance is high due to the large absolute difference between the final components.

The Cosine similarity is moderately high because the original vectors share a somewhat similar positive direction.

The Pearson correlation is low because, after centering, the values ​​change in a poorly coordinated manner.

The Jaccard index reflects the proportion of shared elements and is calculated using different data.

These numerical results cannot be compared directly because Euclidean is an unbounded distance, while cosine, Jaccard, and Pearson are bounded similarity measures. Jaccard is also calculated from different input data.

## 3.2 D2. Dimensionality Reduction (PCA)

```python
from sklearn.datasets import load_iris

iris_data = load_iris()

df = pd.DataFrame(iris_data.data, columns = iris_data.feature_names)

df['target'] = iris_data.target

df.head()
```

```python
df.shape
```

```python
feature_names = ['sepal length (cm)' , 'sepal width (cm)', 'petal length (cm)' , 'petal width (cm)']

X = df[feature_names]
y = df['target']


print(iris_data.target_names)
print(X.shape, y.shape)
print(df.columns)
print(X.isnull().sum())
```

```python
X.describe()
```

```python
x_np = np.array(X)

mean = np.mean(x_np, axis=0)
std = np.std(x_np, axis=0)
z_score = (x_np - mean) / std

print('Shape:', z_score.shape)
print('First 5 rows:')
print(z_score[:5])

print('Mean after standardization:')
print(np.mean(z_score, axis=0))

print('Std after standardization:')
print(np.std(z_score, axis=0))
```

```python
n = z_score.shape[0]

z_score_t = z_score.T

cov_matrix = (z_score_t @ z_score) / (n - 1)

print(cov_matrix.shape)

cov_matrix.round(3)
#cov_matrix[row, column], 
#0 sepal length
#1 - sepal width
#2 - petal length
#3 - petal width
```

petal length и petal width: 0.969 — powerfull relationship

sepal length и petal length: 0.878;

sepal length и petal width: 0.823;

sepal width и petal length: -0.431 — negative relationship

sepal length и sepal width: -0.118 — weak relationship

```python
eigenvalues, eigenvectors = np.linalg.eigh(cov_matrix)

print('Eigenvalues:')
print(eigenvalues)

print('Eigenvectors:')
print(eigenvectors)

print(eigenvalues.shape)
print(eigenvectors.shape)
```

```python
sorted_indices = np.argsort(eigenvalues)[::-1]

sorted_eigenvalues = eigenvalues[sorted_indices]

sorted_eigenvectors = eigenvectors[:,sorted_indices]

top_two_eigenvalues = sorted_eigenvalues[:2]

top_two_eigenvectors = sorted_eigenvectors[:,:2]

print(top_two_eigenvalues)
print(top_two_eigenvectors.shape)
```

```python
sum_of_eigenvalues = np.sum(sorted_eigenvalues)

res = top_two_eigenvalues / sum_of_eigenvalues

for items in res:
    print(f'PC = {items:.4f}')

print(f'Their sum = {np.sum(res):.4f}')
```

```python
pca_data = z_score @ top_two_eigenvectors

print(pca_data.shape)
print(pca_data[:5])
```

```python
pca_dataframe = pd.DataFrame(pca_data,columns=['PC1', 'PC2'],index=df.index)

pca_dataframe['target'] = y

print(pca_dataframe.shape)
pca_dataframe.head()
```

```python
plt.figure(figsize=(8, 6))

for target_value, target_name in enumerate(iris_data.target_names):
    mask = pca_dataframe['target'] == target_value

    plt.scatter(pca_dataframe.loc[mask, 'PC1'],pca_dataframe.loc[mask, 'PC2'],label=target_name,alpha=0.7)

plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')
plt.title('PCA Projection of the Iris Dataset')
plt.legend()
plt.grid(alpha=0.3)
plt.show()
```

### Explanation of PCA Steps

1. **Data preparation:** The four numerical Iris features were separated from the target. The target was not used to calculate PCA; it was only used later to colour the classes in the visualization.

2. **Standardization:** Each feature was transformed to have a mean close to 0 and a standard deviation close to 1. This prevents features with larger scales from dominating PCA.

3. **Covariance matrix:** The covariance matrix shows how the standardized features vary together. For example, petal length and petal width have a strong positive relationship.

4. **Eigenvalues and eigenvectors:** Eigenvectors define the new directions of the principal components. Eigenvalues represent the amount of variance captured by each corresponding eigenvector. A larger eigenvalue means that the component preserves more information from the original data.

5. **Selection of components:** The eigenvalues and eigenvectors were sorted from largest to smallest. The two eigenvectors associated with the two largest eigenvalues were selected as PC1 and PC2.

6. **Projection:** The standardized data were multiplied by the two selected eigenvectors. This transformed the dataset from four original features into two principal components.

7. **Explained variance:** PC1 explains approximately 72.96% of the total variance and PC2 explains approximately 22.85%. Together, they preserve approximately 95.81% of the original variance.

### PCA Interpretation

The original Iris dataset contained four numerical features, which were reduced to two principal components. PC1 explains approximately 72.96% of the total variance, while PC2 explains approximately 22.85%. Together, they preserve approximately 95.81% of the variance.

The PCA plot shows that the Setosa class is clearly separated from the other two classes, mainly along PC1. Versicolor and Virginica form partially distinct groups, although some overlap remains. This means that two principal components preserve most of the important structure of the original four-dimensional dataset, but some information is still lost during dimensionality reduction.

PCA makes the dataset easier to visualize and can reduce computational complexity, but the new components are combinations of the original features and are therefore less interpretable.

```python
for features in distribution_features:

    plt.figure(figsize=(8,5))

    upper_limit = dat[features].quantile(0.99)


    for data_name, data_items in datasets.items():
        plt.hist(data_items[features].dropna(), bins=30, range = (dat[features].min(), upper_limit), density=True, alpha=0.3, label=data_name)



    plt.xlabel(features)
    plt.ylabel('Density')
    plt.title(f'Distribution comparison for {features}')
    plt.legend()
    plt.show()
```

Which sampling method preserved distributions best - Stratified sample, because he almost showed the same distribution in histogram and same mean as original dataset and almost same median.

Impact on outlier detection - sampling can also affect outlier detection. Rare extreme values may not be selected in a smaller random or stratified sample, which can reduce the variance and change the calculated IQR boundaries. Histograms can reveal long tails, while boxplots and the IQR method are more appropriate for identifying individual outliers.

Recommendations for real-world projects - it depends on what goal you want to achieve, Cluster good - when we are talking about georaphicallly distributed population and cheking all of them is expensive.
